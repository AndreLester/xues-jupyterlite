# Optimum Resolution for a GitHub Profile Picture

## Summary

The optimum resolution for a GitHub profile picture is **500 × 500 pixels**.

## Details

| Property | Recommendation |
|---|---|
| **Dimensions** | 500 × 500 px (square) |
| **Maximum file size** | 1 MB |
| **Supported formats** | PNG, JPG, GIF |

### Why 500 × 500 px?

GitHub displays profile pictures at many different sizes across its UI:

- **20 × 20 px** – inline commit/comment avatars
- **40 × 40 px** – activity-feed and pull-request views
- **48 × 48 px** – repository contributor lists
- **96 × 96 px** – profile summary cards
- **240 × 240 px** – full profile page

Uploading an image that is at least **500 × 500 px** ensures it looks sharp at every size without being needlessly large.

### Tips

- Use a **square** image so GitHub does not crop important parts of the picture.
- Keep the file under **1 MB** (GitHub's upload limit).
- **PNG** is preferred for logos or artwork with transparent backgrounds; **JPG** is fine for photographs.
